# Respuesta 67

```python
def dividir_numeros(mensaje):
    while True:
        try:
            numero = int(input(mensaje))
            return numero
        except ValueError:
            print("Error: No ha ingresado un numero valido. Use solo enteros.")
    

print("-" * 50)
print("TRY/EXCEPT")
print("-" * 50)

numero1=dividir_numeros("Ingrese el primer numero: ")
numero2=dividir_numeros("Ingrese el segundo numero: ")

try:
    resultado = numero1 / numero2
    print(f"El resultado de la division es: {resultado}")
except ZeroDivisionError:
    print("Error: No se puede dividir entre cero.")
```

### Desafío 68: 
Crea un programa que tome una lista de valores y realice operaciones matemáticas sobre ellos. Implementa el manejo de varias excepciones comunes como ZeroDivisionError, TypeError, y ValueError.

```python
def realizar_operaciones_matematicas(lista_valores):
    """
    Tomamos una lista de valores e intentamos dividirlos
    """
    print("=== EJECUTANDO DIVISIÓN CON CADA VALOR===")
    # este for provará cada valor con el proceso
    for valor in lista_valores:
        try:
            # Provamos convertir el valor a entero (ValueError si es texto)
            numero = int(valor)
            
            # dividimos 100 entre el número, si es 0 nos dará error de división por cero
            resultado = 100 / numero
            
            print(f"División: 100 / {numero} = {resultado}")
            
        except ValueError:
            print(f" [ValueError]: El valor '{valor}' no se pudo convertir a un número válido.")
            
        except ZeroDivisionError:
            print(f" [ZeroDivisionError]: No se puede dividir entre cero (el valor fue '{valor}').")
            
        except TypeError:
            # aqui mostramos el tipo de dato para que el usiario sepa que valor es
            print(f" [TypeError]: El tipo de dato '{type(valor).__name__}' no es compatible con la operación.")
            
        except Exception as e:
            # Una red de seguridad general para cualquier otro error inesperado
            print(f" [Error inesperado]: {type(e).__name__} - {e}")

# --- PRUEBA 
# Colocamos toda una suerte de datos para probar el programa
valores_prueba = [10, "20", "hola", 0, "0", "5", [1, 2]]

realizar_operaciones_matematicas(valores_prueba)
```

```python
import math
numero_factorial = 0

def factorial(numero_factorial):
    try:
        numero=int(numero_factorial)
        resultado = math.factorial(numero)
        print("El factorial de ", numero, " es ", resultado)
    except ValueError:
        print("Error: el numero no puede ser negativo en un factorial")
    except TypeError:
        print("Error: el numero debe ser entero")
    except OverflowError:
        print("el numero es muy grande para hacer el factorial")
    return True
el_numero = input("ingrese un numero para el factorial")
factorial(el_numero)
```

```python
import math      # libreria de python para operaciones matematicas
class NegativeNumberError (Exception):
    # se hereda de Exception para crear una clase de error
    #personalizada
    # el siguiente comentario en ("""") es para documentar la clase y explicar su uso
    """Clase que hereda de Exception y maneja mensajes de error de 
    raiz cuadrada de numeros negativos."""
    def __init__(self, mensaje="ERROR: No se puede calcular la raíz cuadrada de un número negativo."):
        self.mensaje=mensaje
        super().__init__(self.mensaje)
def calcular_raiz_cuadrada(numero): # función que calcula la raíz cuadrada de un número
    if numero < 0:
    # "raise" muestra el mensaje de error definido en la clase NegativeNumberError, deteniendo
    # la ejecución del programa, evitando que siga funcionando con datos posiblemente corruptos.
        raise NegativeNumberError() 
    else:
        return math.sqrt(numero) # formula para raiz cuadrada, usando la libreria math de python
print("-" * 50)
print("VAMOS A CALCULAR RAICES CUADRADAS")
print("-" * 50)
numero1=int(input("Ingrese un numero para calcular su raiz cuadrada: "))
try:
    resultado = calcular_raiz_cuadrada(numero1) # llamada a función
    print(f"La raiz cuadrada de {numero1} es: {resultado}")
# se puede usar el siguente código con la variable "e" para mostrar el mensaje de error,
#  pero en este caso no es necesario
# except NegativeNumberError as e:
#    print(e)
except NegativeNumberError :
    print(NegativeNumberError().mensaje)

```
