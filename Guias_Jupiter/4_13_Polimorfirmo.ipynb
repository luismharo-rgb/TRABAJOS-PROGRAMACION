# Polimorfismo

El polimorfismo es otro concepto fundamental en la Programación Orientada a Objetos que permite que un objeto pueda adoptar múltiples formas. En el contexto de la herencia, el polimorfismo permite que un objeto de la subclase pueda ser tratado como un objeto de la clase base, lo que facilita la escritura de código más genérico y reutilizable.

## Ejemplo de polimorfismo en la biblioteca

Vamos a ver cómo podemos aplicar el polimorfismo en nuestro ejemplo de la biblioteca. Supongamos que queremos tener una función que imprima la información de cualquier autor, sin importar si es un Autor, un Escritor, un EscritorAcademico, etc.

```python
def imprimir_informacion_autor(autor):
    print("Nombre:", autor.nombre)
    if isinstance(autor, Escritor):
        print("Género Literario:", autor.genero)
    if isinstance(autor, Academico):
        print("Universidad:", autor.universidad)
```

Ahora, podemos pasar cualquier objeto de una clase que herede de Autor a esta función, y se imprimirá la información correspondiente:

```python
autor = Autor("Julio Cortázar")
escritor_academico = EscritorAcademico("Umberto Eco", "Novela Histórica", "Universidad de Bolonia")

imprimir_informacion_autor(autor)
imprimir_informacion_autor(escritor_academico)
```

## Sobrescritura de métodos

El polimorfismo también nos permite sobrescribir métodos en las subclases. Por ejemplo, podríamos tener un método informacion() en la clase Autor y sobrescribirlo en las subclases para que devuelva información adicional:

```python
class Autor:
    def __init__(self, nombre):
        self.nombre = nombre
    
    def informacion(self):
        return f"Nombre: {self.nombre}"

class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero
    
    def informacion(self):
        return f"{super().informacion()} - Género Literario: {self.genero}"

# Instanciamos un objeto de la clase Escritor para Mario Benedetti
escritor = Escritor("Mario Benedetti", "Realismo Social")
print(escritor.informacion())
```

El polimorfismo, junto con la herencia, nos permite escribir código más flexible y reutilizable, al permitirnos tratar objetos de subclases como si fueran objetos de la clase base y sobrescribir métodos para añadir funcionalidades específicas a las subclases.

## Ejemplos adicionales de polimorfismo
### Polimorfismo con métodos de clase

```python
class Animal:
    def sonido(self):
        return "Algunos animales hacen sonidos"

class Perro(Animal):
    def sonido(self):
        return "Guau Guau"

class Gato(Animal):
    def sonido(self):
        return "Miau Miau"
    
class Loro(Animal):
    def sonido(self):
        return "Prr Prr"

animales = [Perro(), Gato(), Loro(), Animal()]

for animal in animales:
    print(animal.sonido())
```

# Respuesta 62

```python
# --- Clase Base ---
class Musico:
    """esta clase genera los datos generales para un músico."""
    def __init__(self, nombre):
        self.nombre = nombre

    def instrumento(self):
        """este es el método que será sobreescrito por las subclases."""
        print(f"{self.nombre} es un músico.")


# --- Subclases que sobrescriben el método ---
class Guitarrista(Musico):
    """Subclase Guitarrista. Hereda de Músico, la 
     "instrumento" sobreescribe la clase "instrumento" de Musico"""
    def instrumento(self):
        print(f" {self.nombre} es guitarrista y toca la guitarra eléctrica.")


class Baterista(Musico):
    """Subclase Baterista.Hereda de Músico, la 
     "instrumento" sobreescribe la clase "instrumento" de Musico"""
    def instrumento(self):
        print(f" {self.nombre} es baterista y toca la batería acústica.")


# --- Polimorfismo ---
# Datos - objetos de distintas clases
musico_generico = Musico("Carlos")
guitarrista = Guitarrista("Lucía")
baterista = Baterista("Mateo")

# Creamos una lista con diferentes músicos 
banda = [musico_generico, guitarrista, baterista] # datos de distinas clases

print("=== BANDA 'POLI-BANDA' ===")
for m in banda:
    # Llamamos al mismo método .instrumento() en cada objeto
    # Python sabe automáticamente qué versión del método ejecutar según la clase del objeto
    m.instrumento()
```

```python
import math

class Figura:
    # esta clase no se instancia, es solo un "molde" para las demás
    # así podemos crear una función area() que sea llamada con el mismo nombre
    # pero funcione diferente en cada clase hija.
    def area(self):
       pass

class Cuadrado(Figura):

    def __init__(self, lado):
        # usar el nombre del método area() y sobre escribirlo en cada clase hija
        # Al no tener datos, la clase padre no hace nada, pero si tuvieramos datos, los heredaría
        self.lado =lado

    def area(self):
        return self.lado ** 2
    
class Circulo(Figura):
    def __init__(self, radio):
        self.radio = radio

    def area(self):
        return math.pi * (self.radio ** 2)      

calculo_cuadrado= Cuadrado(5)
calculo_circulo= Circulo(3)

figuras=[calculo_cuadrado, calculo_circulo]

print(" === ÁREAS DE LAS FIGURAS GEOMÉTRICAS ===")
print("-" * 42)
for figura in figuras:
# el type lo que hace es mostrar el nombre de la clase a la que pertenece el objeto
# por eso llama a "__name__" 
    print()
    print(f"El área de la figura {type(figura).__name__} es: {figura.area()}")

        
        
```

### Desafío 66: Polimorfismo en operaciones matemáticas
En este desafío, aplicarás el polimorfismo para realizar diferentes operaciones matemáticas. Deberás crear una clase base Operacion con un método resultado y dos subclases Suma y Multiplicacion que sobrescriban este método para realizar las operaciones correspondientes.

```python
class Operacion:
    # Esta clase no se instancia, es solo un "molde" (interfaz) para las demás.
    # Así podemos crear un método ejecucion() que sea llamado con el mismo nombre
    # pero funcione diferente en cada clase hija.
    def ejecucion(self):
        pass


class Suma(Operacion):
    def __init__(self, numero1, numero2):
        self.numero1 = numero1
        self.numero2 = numero2

    # Sobrescribimos ejecucion() usando los atributos self del objeto
    def ejecucion(self):
        return self.numero1 + self.numero2


class Resta(Operacion):
    def __init__(self, numero1, numero2):
        self.numero1 = numero1
        self.numero2 = numero2

    # Sobrescribimos ejecucion() usando los atributos self del objeto
    def ejecucion(self):
        return self.numero1 - self.numero2


# --- Datos de prueba ---
numero1 = 2
numero2 = 4

calculo_suma = Suma(numero1, numero2)
calculo_resta = Resta(numero1, numero2)

# Lista que almacena las instancias (polimorfismo)
ejercicios = [calculo_suma, calculo_resta]

print("=== EJERCICIOS DE SUMA Y RESTA ===")
print("-" * 42)

for ejercicio in ejercicios:
    # 1. type(ejercicio).__name__ obtiene el nombre de la clase ('Suma' o 'Resta')
    # 2. ejercicio.ejecucion() llama al método correspondiente a cada subclase
    nombre_clase = type(ejercicio).__name__
    resultado = ejercicio.ejecucion()
    print(f"El cálculo del ejercicio {nombre_clase} es: {resultado}")
```
